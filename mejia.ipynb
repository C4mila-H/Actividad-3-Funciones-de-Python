## 1. Método `.count()`
*   **¿Para qué sirve?** Cuenta cuántas veces aparece una letra, una palabra o un número específico dentro de un texto o una lista.
*   **Sintaxis básica:** `texto.count("elemento_a_buscar")`



```python
# Ejemplo de .count: Contar cuántas veces se repite una palabra en una bitácora ambiental
registro = "Muestra de agua tomada en el río. El río presenta niveles altos de turbiedad."
repeticiones = registro.count("río")

print("La palabra 'río' aparece:", repeticiones, "veces.")
```

## 2. Método `.find()`
*   **¿Para qué sirve?** Busca una palabra o letra dentro de un texto y te devuelve la **posición exacta (índice)** donde comienza. Si no encuentra la palabra, devuelve `-1`.
*   **Sintaxis básica:** `texto.find("palabra_a_buscar")`



```python
# Ejemplo de .find: Buscar la ubicación de un término clave
correo_alerta = "Alerta: Se detectó presencia de plomo en el sector norte"
posicion = correo_alerta.find("plomo")

print("La palabra 'plomo' se encuentra en la posición de carácter número:", posicion)

    
```

## 3. Función `id()`
*   **¿Para qué sirve?** Devuelve un número de identificación único para cualquier objeto o variable en Python. Es como la "cédula" o huella digital que tiene cada dato dentro de la memoria del computador mientras el programa corre.
*   **Sintaxis básica:** `id(variable)`



```python
# Ejemplo de id: Ver la identidad única de dos variables en memoria
estacion_1 = "Facatativá"
estacion_2 = "Apartadó"

print("Identidad única de la estación 1:", id(estacion_1))
print("Identidad única de la estación 2:", id(estacion_2))

```

## 4. Método `.upper()`
*   **¿Para qué sirve?** Convierte todas las letras de una cadena de texto (string) a **mayúsculas**.
*   **Sintaxis básica:** `texto.upper()`


```python
# Ejemplo de .upper: Estandarizar una respuesta
respuesta_usuario = "aceptar"
en_mayusculas = respuesta_usuario.upper()

print("Texto transformado:", en_mayusculas)
```

## 5. Método `.lower()`
*   **¿Para qué sirve?** Convierte todas las letras de una cadena de texto a **minúsculas**. Es sumamente útil para comparar textos sin importar cómo los escribió el usuario.
*   **Sintaxis básica:** `texto.lower()`


```python
# Ejemplo de .lower: Validar un correo electrónico
correo = "Usuario_VIP@Gmail.com"
correo_limpio = correo.lower()

print("Correo en minúsculas:", correo_limpio)
```

## 6. Método `.replace()`
*   **¿Para qué sirve?** Busca una palabra o un carácter específico dentro de un texto y lo **reemplaza** por otro texto nuevo.
*   **Sintaxis básica:** `texto.replace("viejo", "nuevo")`


```python
# Ejemplo de .replace: Corregir una palabra en una frase
frase = "Me encanta programar en Java"
frase_corregida = frase.replace("Java", "Python")

print("Frase corregida:", frase_corregida)
```

## 7. Método `.split()`
*   **¿Para qué sirve?** Corta o **divide** una cadena de texto en varias partes utilizando un separador (por defecto, los espacios) y guarda el resultado dentro de una lista.
*   **Sintaxis básica:** `texto.split("separador")`


```python
# Ejemplo de .split: Separar una lista de compras escrita en texto
compras = "manzana,pera,plátano,fresa"
lista_compras = compras.split(",")

print("Texto original:", compras)
print("Lista resultante de Python:", lista_compras)
```

## 8. Método `.join()`
*   **¿Para qué sirve?** Es el proceso inverso a `.split()`. Toma una lista de textos y los **une** todos en una sola cadena de texto, intercalando el separador que tú elijas entre cada elemento.
*   **Sintaxis básica:** `"separador".join(lista)`


```python
# Ejemplo de .join: Unir palabras para formar una fecha limpia
componentes_fecha = ["04", "11", "2026"]
fecha_unida = "/".join(componentes_fecha)

print("Fecha formateada:", fecha_unida)
```

## 9. Método `.append()`
*   **¿Para qué sirve?** Añade un nuevo elemento al **final** de una lista existente. Modifica la lista directamente sumando el dato que le indiques.
*   **Sintaxis básica:** `lista.append(elemento)`


```python
# Ejemplo de .append: Agregar alumnos a un curso
alumnos = ["Carlos", "Sofía"]
alumnos.append("Laura")

print("Lista de alumnos actualizada:", alumnos)
```

## 10. Función `any()`
*   **¿Para qué sirve?** Revisa una lista o colección de valores booleanos (`True` o `False`) y devuelve `True` si **al menos uno** de los elementos de la lista es verdadero. Si todos son falsos, devuelve `False`.
*   **Sintaxis básica:** `any(lista_de_booleanos)`


```python
# Ejemplo de any: Verificar si hay algún artículo con descuento
descuentos_aplicados = [False, False, True, False]
hay_descuentos = any(descuentos_aplicados)

print("¿Hay algún producto con descuento en la tienda?:", hay_descuentos)
```
