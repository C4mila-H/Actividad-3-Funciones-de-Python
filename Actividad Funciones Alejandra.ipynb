# 1. abs(): ¿Cuántos centímetros se desvía una viga si el diseño pedía 105 y se midió 108.5?

`abs()` devuelve el valor absoluto, es decir, el número sin signo. Sirve para calcular diferencias sin importar si es por exceso o por defecto.

```python
abs(105 - 108.5)
```

# 2. pow(): ¿Cuál es el área de una zapata cuadrada de 3 metros de lado?

`pow(base, exponente)` calcula una potencia, igual que el operador `**`.

```python
pow(3, 2)
```

# 3. divmod(): Tengo 17 bultos de cemento para repartir entre 5 columnas, ¿cuántos recibe cada una y cuántos sobran?

`divmod(a, b)` devuelve el cociente y el residuo de una división.

```python
divmod(17, 5)
```

# 4. sorted(): Ordenar de menor a mayor la resistencia (en MPa) de 4 cilindros de concreto

`sorted()` devuelve una lista nueva ordenada sin modificar la original.

```python
sorted([28, 21, 35, 24])
```

# 5. reversed(): Mostrar al revés las etapas de una obra

`reversed()` recorre una secuencia de atrás hacia adelante; se convierte a lista para verla.

```python
list(reversed(["Cimentación", "Estructura", "Acabados"]))
```

# 6. enumerate(): Numerar las etapas de construcción de una obra

`enumerate()` entrega el número de posición y el valor al recorrer una lista.

```python
for numero, etapa in enumerate(["Cimentación", "Estructura", "Acabados"], start=1):
    print(numero, etapa)
```

# 7. zip(): Unir cada elemento estructural con su longitud en metros

`zip()` combina varias listas elemento a elemento.

```python
elementos = ["Viga", "Columna", "Losa"]
longitudes = [6.0, 3.5, 4.2]
list(zip(elementos, longitudes))
```

# 8. str(): Convertir la altura de una columna (3.5) a texto para armar una frase

`str()` convierte cualquier valor a texto; es necesario para unir números con palabras.

```python
"La columna mide " + str(3.5) + " metros"
```

# 9. bool(): ¿Se registró el nombre de la obra? (con nombre y sin nombre)

`bool()` devuelve True o False; un texto vacío es False y uno con contenido es True.

```python
print(bool("Edificio Central"))
print(bool(""))
```

# 10. range(): Mostrar los 7 días de curado del concreto

`range(inicio, fin)` genera números seguidos; el valor final no se incluye.

```python
list(range(1, 8))
```
